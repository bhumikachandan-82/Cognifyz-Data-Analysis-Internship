```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("C:/Users/Bhumika Chandan/Downloads/Dataset .csv")

print("Dataset loaded successfully!")
```

```python
city_counts = (
    df.groupby("City")["Restaurant ID"]
    .nunique()
    .sort_values(ascending=False)
)

print("Top 10 Cities by Restaurant Count:")
print(city_counts.head(10))
```

```python
highest_restaurant_city = city_counts.idxmax()
highest_restaurant_count = city_counts.max()

print("City with highest restaurants:", highest_restaurant_city)
print("Number of restaurants:", highest_restaurant_count)
```

```python
city_ratings = (
    df[df["Aggregate rating"] > 0]
    .groupby("City")["Aggregate rating"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Rating by City:")
print(city_ratings)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=city_counts.head(10).index,
    y=city_counts.head(10).values,
    color="teal",
    width=0.6
)

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("Top_10_Cities.png", dpi=300)
plt.show()
```

```python
result_city = pd.DataFrame({
    "City": city_ratings.index,
    "Average Rating": city_ratings.values
})

result_city.to_csv("City_Average_Ratings.csv", index=False)

city_counts.rename("Restaurant Count").to_csv(
    "City_Restaurant_Counts.csv"
)

print("Task 2 completed successfully!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("C:/Users/Bhumika Chandan/Downloads/Dataset .csv")

print("Dataset loaded successfully!")
```

```python
print("Unique Price Ranges:")
print(df["Price range"].unique())
```

```python
price_counts = df["Price range"].value_counts().sort_index()

print("Restaurant Count by Price Range:")
print(price_counts)
```

```python
total_restaurants = price_counts.sum()

price_percentage = (price_counts / total_restaurants) * 100

print("Percentage by Price Range:")
print(price_percentage.round(2))
```

```python
result = pd.DataFrame({
    "Price Range": price_counts.index,
    "Restaurant Count": price_counts.values,
    "Percentage": price_percentage.values.round(2)
})

display(result)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=price_counts.index,
    y=price_counts.values,
    palette=["#264653", "#2A9D8F", "#E9C46A", "#E76F51"],
    hue=price_counts.index,
    legend=False,
    width=0.4
)

plt.title("Distribution of Restaurant Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")

plt.tight_layout()
plt.savefig("Price_Range_Distribution.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=price_counts.index,
    y=price_counts.values,
    palette=["#264653", "#2A9D8F", "#E9C46A", "#E76F51"],
    hue=price_counts.index,
    legend=False,
    width=0.4
)

for i, value in enumerate(price_counts.values):
    percentage = price_percentage.iloc[i]

    ax.text(
        i,
        value + 50,
        f"{percentage:.2f}%",
        ha="center"
    )

plt.title("Restaurant Price Range Distribution (%)")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")

plt.tight_layout()
plt.savefig("Price_Range_Percentage_Chart.png", dpi=300)
plt.show()
```
